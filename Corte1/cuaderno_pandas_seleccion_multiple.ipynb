# Selección de Múltiples Columnas en Pandas

### ¿Cómo seleccionar más de una columna a la vez?
A diferencia de extraer una sola columna (que devuelve una **Series**), al seleccionar **dos o más columnas** en **Pandas** el resultado siempre será un nuevo **DataFrame** subconjunto.

- **Sintaxis de Corchetes Dobles (`df[['Col1', 'Col2']]`):** Se utiliza un par interno de corchetes para definir la **lista de nombres de columnas** que se desea extraer, y un par externo de corchetes para indexar el DataFrame.
- **Orden de las columnas:** El nuevo DataFrame mantendrá las columnas exactamente en el orden en que se especifiquen dentro de la lista.

### Línea 1
**Código:** `import pandas as pd`

**Explicación:** Importa la biblioteca Pandas con el alias estándar 'pd' para manipular DataFrames.

```python
import pandas as pd
```

### Línea 2
**Código:** `nba = pd.read_excel("Nba Teams.xlsx")`

**Explicación:** Carga el archivo Excel 'Nba Teams.xlsx' mediante pd.read_excel() y guarda el DataFrame en la variable 'nba'.

```python
nba = pd.read_excel("Nba Teams.xlsx")
```

### Línea 3
**Código:** `nba.head()`

**Explicación:** Visualiza en la consola las primeras 5 filas del DataFrame 'nba' usando el método .head().

```python
nba.head()
```

### Línea 4
**Código:** `nba[["Name", "Team"]]`

**Explicación:** Selecciona múltiples columnas ('Name' y 'Team') pasando una lista de nombres dentro de los corchetes dobles [[...]], devolviendo un nuevo DataFrame filtrado con solo esas dos columnas.

```python
nba[["Name", "Team"]]
```

### Línea 5
**Código:** `nba[["Name", "Team", "Salary"]]`

**Explicación:** Selecciona un subconjunto de tres columnas ('Name', 'Team' y 'Salary') utilizando corchetes dobles [[...]], retornando un DataFrame de tres columnas.

```python
nba[["Name", "Team", "Salary"]]
```
