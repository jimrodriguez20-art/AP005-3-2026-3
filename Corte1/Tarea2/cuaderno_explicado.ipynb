# Guía de Diccionarios en Python
Explicación línea por línea con bloques Markdown individuales adjuntos a cada instrucción de código.

### Línea 1
**Sintaxis y operación:** Declara una lista de cadenas de texto (`str`) con los nombres de las canciones.
**Importancia y relación:** Define la secuencia de datos base que se utilizarán como las claves en el diccionario de reproducciones.

```python
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]
```

### Línea 2
**Sintaxis y operación:** Declara una lista de números enteros (`int`) con la cantidad de reproducciones.
**Importancia y relación:** Establece los valores numéricos correspondientes que se asociarán a cada canción.

```python
playcounts = [78, 29, 44, 21, 89, 5]
```

### Línea 3
**Sintaxis y operación:** Genera un diccionario mediante comprensión emparejando las listas `songs` y `playcounts` con la función `zip()`.
**Importancia y relación:** Construye de forma dinámica la estructura clave-valor relacionando cada canción con su conteo de reproducciones.

```python
plays = {key:value for key, value in zip(songs, playcounts)}
```

### Línea 4
**Sintaxis y operación:** Muestra en la consola el contenido del diccionario `plays` usando la función `print()`.
**Importancia y relación:** Permite verificar la correcta construcción del diccionario inicial.

```python
print(plays)
```

### Línea 5
**Sintaxis y operación:** Invoca el método `.update()` para añadir la clave `"Purple Haze"` con valor `1`.
**Importancia y relación:** Incorpora un elemento nuevo al diccionario existente sin sobrescribir el resto de datos.

```python
plays.update({"Purple Haze": 1})
```

### Línea 6
**Sintaxis y operación:** Modifica el valor asociado a la clave existente `"Respect"` actualizándolo a `94` mediante `.update()`.
**Importancia y relación:** Demuestra la mutabilidad de los valores en los diccionarios actualizando una métrica previa.

```python
plays.update({"Respect": 94})
```

### Línea 7
**Sintaxis y operación:** Imprime en consola el texto `"After: "` junto al estado del diccionario `plays` modificado.
**Importancia y relación:** Muestra los resultados finales en pantalla tras la inserción y modificación de datos.

```python
print("After: ", plays)
```

### Línea 8
**Sintaxis y operación:** Declara el diccionario `library` que contiene a `plays` como valor de `"The Best Songs"` y un diccionario vacío `{}` en `"Sunday Feelings"`.
**Importancia y relación:** Construye una estructura de datos jerárquica y anidada (*nested dictionary*).

```python
library = {"The Best Songs": plays, "Sunday Feelings": {}}
```

### Línea 9
**Sintaxis y operación:** Muestra en pantalla el diccionario completo `library`.
**Importancia y relación:** Confirma la integración correcta del diccionario anidado.

```python
print(library)
```
