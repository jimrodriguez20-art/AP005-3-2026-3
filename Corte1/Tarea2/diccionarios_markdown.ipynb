# Módulo 2: Comprensión de Diccionarios y Estructuras Anidadas
Esta sección demuestra cómo empaquetar secuencias paralelas con `zip()`, generar diccionarios dinámicamente con *dict comprehensions* y construir estructuras de datos anidadas.

### Explicación del Código - Módulo 2
| Línea de Código | ¿Qué hace? (Sintaxis y Operación) | Importancia y Relación con el Código |
| :--- | :--- | :--- |
| `songs = ["Like a Rolling Stone", ...]` | Define una lista con los títulos de las canciones. | Establece la secuencia de datos para las claves primarias. |
| `playcounts = [78, 29, 44, 21, 89, 5]` | Define una lista de enteros con el conteo de reproducciones. | Proporciona los valores cuantitativos para cada canción. |
| `plays = {key:value for key, value in zip(songs, playcounts)}` | Genera un diccionario mapeando `songs` y `playcounts` con `zip()` y comprensión. | Construye la estructura clave-valor principal combinando las dos listas. |
| `print(plays)` | Imprime el diccionario `plays` en la consola. | Muestra el estado inicial del diccionario de reproducciones. |
| `plays.update({"Purple Haze": 1})` | Inserta el par `"Purple Haze": 1` usando `.update()`. | Añade una nueva entrada al diccionario existente. |
| `plays.update({"Respect": 94})` | Modifica el valor de la clave `"Respect"` a `94`. | Actualiza un dato existente demostrando la mutabilidad de los valores. |
| `print("After: ", plays)` | Imprime el diccionario `plays` tras las modificaciones. | Despliega los cambios aplicados en la estructura. |
| `library = {"The Best Songs": plays, "Sunday Feelings": {}}` | Declara el diccionario anidado `library` incorporando `plays`. | Demuestra cómo construir estructuras de datos jerárquicas y compuestas. |
| `print(library)` | Imprime la estructura completa de `library`. | Muestra el resultado final del diccionario anidado. |

```python
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]
playcounts = [78, 29, 44, 21, 89, 5]
plays = {key:value for key, value in zip(songs, playcounts)}
print(plays)
plays.update({"Purple Haze": 1})
plays.update({"Respect": 94})
print("After: ", plays)
library = {"The Best Songs": plays, "Sunday Feelings": {}}
print(library)
```
