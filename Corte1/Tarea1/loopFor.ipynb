# Guía Educativa: Control de Flujo en Iteraciones con `for`, `continue` y el Módulo `time` (`loopFor.py`)

Este notebook presenta una explicación estructurada sobre cómo iterar sobre secuencias (como cadenas de texto) utilizando el bucle `for`, cómo modificar la secuencia normal de iteración mediante la sentencia de control `continue` y cómo pausar temporalmente la ejecución del programa utilizando la librería estándar `time`.

## 1. Importación de Módulos y Definición de Datos

En esta primera sección se importa el módulo `time` perteneciente a la biblioteca estándar de Python, el cual proporciona funciones para trabajar con tiempos e intervalos. Posteriormente, se define una variable de tipo cadena de texto (`str`) sobre la cual se iterará más adelante.

```python
# Importa el módulo 'time' para acceder a funciones de manejo de tiempo y retardos en la ejecución
import time

# Define una variable de tipo cadena de texto (str) que servirá como secuencia iterarle
cadena = 'Python'
```

## 2. Iteración con Bucle `for`, Salto de Iteración con `continue` y Pausa de Ejecución

En este bloque se utiliza un bucle `for` para recorrer individualmente cada carácter de la variable `cadena`. 

Dentro del cuerpo del bucle se incluye una estructura condicional (`if`) que verifica si el carácter actual es igual a `'t'`. Si la condición es verdadera, la sentencia `continue` interrumpe de inmediato la iteración actual y pasa a la siguiente, omitiendo las líneas de código posteriores (`print` y `time.sleep`). Para el resto de los caracteres, el carácter se imprime en pantalla y el programa se pausa durante 1 segundo mediante `time.sleep(1)`.

```python
# Inicia el bucle 'for' para iterar carácter por carácter sobre la cadena 'Python'
for letra in cadena:
    # Evalúa si la letra de la iteración actual es idéntica al carácter 't'
    if letra == 't':
        # Salta el resto del código dentro del bucle para esta iteración y avanza al siguiente elemento
        continue
    
    # Muestra en pantalla el carácter correspondiente a la iteración actual
    print(letra)
    
    # Pausa la ejecución del programa durante 1.0 segundo antes de continuar con el bucle
    time.sleep(1)
```

## Conclusión

En esta guía prcatica se abordaron tres conceptos clave de la programación en Python:

1. **Iteración sobre cadenas de texto:** Las cadenas (`str`) en Python son secuencias iterables, lo que permite a un bucle `for` acceder carácter por carácter sin necesidad de gestionar índices manualmente.
2. **Sentencia de control `continue`:** Permite omitir selectivamente la ejecución del bloque restante dentro de un bucle para una iteración específica cuando se cumple una condición determinada, sin terminar el bucle por completo (a diferencia de `break`).
3. **Módulo `time`:** La función `time.sleep(segundos)` introduce retardo de tiempo intencional en la secuencia de comandos, útil para simular procesos en tiempo real, rate limiting o animaciones simples en la consola.
