# Números Divisibles entre 12 (`divisible12.py`)

Este notebook explica un script corto en Python cuyo objetivo es recorrer un rango de números e imprimir únicamente aquellos que son **divisibles entre 12**.

## Revisión previa del código

Se revisó el script original y **no se encontraron errores**: es sintáctica y lógicamente correcto.

## Nota sobre la estructura de este notebook

El script consiste en un único `for` con un `if`/`continue` anidados, es decir, **una sola sentencia compuesta**. A diferencia de un script con varias instrucciones independientes, este no puede dividirse en varias celdas de código sin romper su ejecución (por ejemplo, una celda que solo contenga `for i in range(100, 301):` sin su cuerpo indentado produciría un `SyntaxError`).

Por esta razón, la explicación conceptual sí se divide en varias subsecciones (el recorrido del rango, el filtro con el operador módulo, la impresión del resultado), pero el código en sí se presenta en **una sola celda ejecutable**, con un comentario para cada línea.

---


## 1. ¿Qué hace este script?

El script recorre los números enteros de un rango determinado y, para cada uno, verifica si es múltiplo de 12. Si lo es, lo imprime; si no lo es, simplemente lo salta y continúa con el siguiente número.

A continuación se explica cada parte del código antes de mostrarlo completo.

### Recorrido del rango de números (`for` + `range`)

`range(100, 301)` genera una secuencia de números enteros que comienza en `100` y termina en `300`. Es importante notar que el segundo argumento de `range()` es **exclusivo**: por eso se usa `301` y no `300`, para que el número `300` sí quede incluido en el recorrido.

El bucle `for i in range(100, 301):` toma, uno por uno, cada número de esa secuencia y lo asigna a la variable `i` para poder evaluarlo dentro del cuerpo del bucle.

### Filtrado con el operador módulo (`%`) y `continue`

El operador módulo (`%`) devuelve el **residuo** de una división entera. Si `i % 12` es igual a `0`, significa que `i` es divisible entre 12 sin dejar residuo (es decir, es múltiplo de 12).

La condición `if (i % 12) != 0:` se cumple precisamente cuando `i` **no** es múltiplo de 12. En ese caso, la instrucción `continue` detiene la iteración actual y salta directamente a la siguiente vuelta del bucle, **sin ejecutar** el resto del código que sigue dentro del `for` (en este caso, el `print(i)`).

### Impresión del resultado (`print`)

Si el número `i` **sí** es múltiplo de 12 (la condición del `if` fue falsa y por lo tanto `continue` no se ejecutó), el flujo del programa llega a la línea `print(i)`, que imprime ese número en la salida.

## 2. Código completo comentado

A continuación se muestra el script completo, con un comentario explicando cada línea.

````python
# Recorre los números enteros desde 100 hasta 300 (300 incluido, porque el límite superior de range() es exclusivo)
for i in range(100, 301):
    # Verifica si "i" NO es divisible entre 12 (el residuo de i % 12 es distinto de 0)
    if (i % 12) != 0:
        # Si no es múltiplo de 12, se salta esta iteración y se pasa directamente al siguiente número
        continue
    # Si el flujo llega aquí, "i" sí es múltiplo de 12, así que se imprime
    print(i)
````

## Conclusión

Este script es un ejemplo sencillo pero representativo de un patrón muy común en Python: **recorrer un rango de valores y filtrar solo los que cumplen cierta condición**.

Los elementos clave que se repasaron son:

- `range(inicio, fin)` genera una secuencia de enteros donde `fin` **no** se incluye, por lo que hay que sumar 1 al límite deseado si se quiere incluirlo (de ahí el `301` para llegar hasta `300`).
- El operador módulo (`%`) permite comprobar divisibilidad: `i % n == 0` es verdadero cuando `i` es múltiplo de `n`.
- La instrucción `continue` interrumpe la iteración actual de un bucle y pasa directamente a la siguiente, sin ejecutar el código que queda debajo dentro del cuerpo del bucle.

Este mismo patrón (recorrer, filtrar con `if`/`continue`, y actuar sobre lo que pasa el filtro) es la base de tareas mucho más complejas de procesamiento de datos en Python.
