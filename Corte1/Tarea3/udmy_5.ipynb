# Selección de Columnas y Objetos Series en Pandas

### ¿Cómo seleccionar columnas en un DataFrame?
En **Pandas**, una columna individual extraída de un DataFrame se denomina **Series** (una estructura de datos unidimensional con etiquetas/índices).

Existen dos formas principales de acceder a una columna:
- **Notación de corchetes (`df['Nombre_Columna']`):** Es la sintaxis estándar y recomendada. Funciona siempre, incluso si el nombre de la columna contiene espacios o caracteres especiales.
- **Notación de punto (`df.Nombre_Columna`):** Permite acceder a la columna como un atributo del DataFrame. Es más corta pero solo funciona si el nombre de la columna es una cadena sin espacios y no entra en conflicto con métodos propios de Pandas (como `.count` o `.shape`).

### Línea 1
**Código:** `import pandas as pd`

**Explicación:** Importa la biblioteca Pandas con el alias estándar 'pd' para habilitar la manipulación de DataFrames y Series.

```python
import pandas as pd
```

### Línea 2
**Código:** `nba = pd.read_excel("Nba Teams.xlsx")`

**Explicación:** Carga el archivo de hoja de cálculo 'Nba Teams.xlsx' mediante pd.read_excel() y almacena el DataFrame en la variable 'nba'.

```python
nba = pd.read_excel("Nba Teams.xlsx")
```

### Línea 3
**Código:** `nba.head()`

**Explicación:** Invoca el método .head() para visualizar en consola las primeras 5 filas del DataFrame 'nba'.

```python
nba.head()
```

### Línea 4
**Código:** `nba["Name"]`

**Explicación:** Selecciona la columna 'Name' del DataFrame 'nba' utilizando la notación de corchetes ['Name'], devolviendo un objeto de tipo Series con los datos de esa columna.

```python
nba["Name"]
```

### Línea 5
**Código:** `nba["Salary"]`

**Explicación:** Selecciona la columna 'Salary' mediante corchetes ['Salary'], retornando la Series que contiene la información salarial.

```python
nba["Salary"]
```

### Línea 6
**Código:** `nba.Name`

**Explicación:** Accede a la columna 'Name' utilizando la notación de punto (.Name) como una propiedad del DataFrame; equivale a nba['Name'] siempre que el nombre de la columna no contenga espacios ni coincida con métodos nativos.

```python
nba.Name
```
