# Lectura e Inspección de Archivos Excel con Pandas

### ¿Cómo leer archivos de Excel (`.xlsx`) en Pandas?
Además de procesar archivos de texto plano como CSV, **Pandas** incluye la función `pd.read_excel()`, diseñada específicamente para importar libros de trabajo de Microsoft Excel (`.xlsx` o `.xls`) directamente a un **DataFrame**.

- **Carga de datos:** `pd.read_excel()` lee por defecto la primera hoja del archivo Excel.
- **Inspección rápida:** El método `.head()` permite previsualizar las primeras filas de la tabla.
- **Diagnóstico de estructura:** El método `.info()` ofrece una descripción detallada de los tipos de columna, facilitando la detección de datos faltantes (*null values*) o inconsistencias.

### Línea 1
**Código:** `import pandas as pd`

**Explicación:** Importa la biblioteca Pandas con el alias estándar 'pd' para manipular conjuntos de datos tabulares.

```python
import pandas as pd
```

### Línea 2
**Código:** `nba = pd.read_excel('Nba Teams.xlsx')`

**Explicación:** Llama a la función pd.read_excel() para cargar el archivo de hojas de cálculo 'Nba Teams.xlsx' y convertirlo en un DataFrame asignado a la variable 'nba'.

```python
nba = pd.read_excel('Nba Teams.xlsx')
```

### Línea 3
**Código:** `nba`

**Explicación:** Muestra la representación completa del DataFrame 'nba' en la salida de la celda de ejecución.

```python
nba
```

### Línea 4
**Código:** `nba.info()`

**Explicación:** Invoca el método .info() para obtener un resumen del DataFrame (número total de filas, columnas, tipos de datos y valores no nulos).

```python
nba.info()
```

### Línea 5
**Código:** `nba.head()`

**Explicación:** Utiliza el método .head() para visualizar en consola las primeras 5 filas del conjunto de datos por defecto.

```python
nba.head()
```

### Línea 6
**Código:** `nba.head(5).info()`

**Explicación:** Encadena el método .info() sobre el subconjunto de las primeras 5 filas devueltas por .head(5) para desplegar el resumen estructural de esa muestra.

```python
nba.head(5).info()
```
