# Restablecimiento del Orden por Índices con `.sort_index()` en Pandas

### ¿Cómo ordenar un DataFrame por sus índices?
Cuando ordenamos un **DataFrame** por sus columnas (usando `.sort_values()`), las filas cambian de posición pero conservan sus **índices de origen**. Esto hace que las etiquetas de las filas queden desordenadas secuencialmente.

- **Propósito de `.sort_index()`:** Permite reordenar el DataFrame según el orden de sus etiquetas de índice (por defecto de forma numérica ascendente).
- **Restauración de secuencia:** Es la herramienta ideal cuando se requiere volver a la disposición u orden cronológico/secuencial original del DataFrame tras haber aplicado filtros u ordenamientos por valores de columna.

### Línea 3
**Código:** `nba.sort_values(["Team", "Name"], na_position = "first", inplace = True)`

**Explicación:** Ordena el DataFrame por las columnas 'Team' y 'Name', posicionando los valores nulos al principio (na_position='first') y aplicando los cambios directamente sobre 'nba' (inplace=True).

```python
nba.sort_values(["Team", "Name"], na_position = "first", inplace = True)
```
