# Filtrado Condicional de Datos y Carga Eficiente en Pandas

### ¿Cómo filtrar filas según condiciones booleanas?
En **Pandas**, el filtrado condicional permite extraer subconjuntos de datos seleccionando filas que cumplen con criterios específicos:

- **Máscaras Booleanas (`df[condicion]`):** Una condición booleana (como `df['gender'] == 'Male'`) devuelve una Series de valores `True` o `False`. Pasar esta condición dentro de los corchetes del DataFrame filtra y retorna únicamente las filas correspondientes a `True`.
- **Operadores de Comparación:** Se utilizan operadores estándar como igualdad (`==`), desigualdad (`!=`), mayor que (`>`) o menor que (`<`).
- **Optimización en Carga con `parse_dates`:** El parámetro `parse_dates=['col1', 'col2']` en `pd.read_csv()` permite convertir columnas de texto a tipo `datetime` de forma automática desde la lectura del archivo.
