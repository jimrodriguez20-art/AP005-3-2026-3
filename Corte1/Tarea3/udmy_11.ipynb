# Imputación y Reemplazo de Datos Nulos con `.fillna()` en Pandas

### ¿Cómo reemplazar valores vacíos (`NaN`) en Pandas?
En el análisis de datos, no siempre es conveniente eliminar filas con datos faltantes. El método `.fillna()` permite **imputar** o reemplazar esos valores vacíos por un valor predeterminado (un número, una cadena o una estadística como la media):

- **Imputación Global (`df.fillna(valor)`):** Reemplaza todos los valores nulos en **todas las columnas** del DataFrame a la vez.
- **Imputación Específica por Columna (`df['Columna'].fillna(valor, inplace=True)`):** Permite aplicar un valor de relleno personalizado para una columna en particular (por ejemplo, `'Missing name'` para nombres o `0` para cantidades cuantitativas) de forma permanente con `inplace=True`.

### Línea 9
**Código:** `nba["Name"].fillna("Missing name", inplace = True)`

**Explicación:** Selecciona la columna 'Name' como Series mediante corchetes simples y reemplaza de forma permanente (inplace=True) todos sus valores nulos por la cadena 'Missing name'.

```python
nba["Name"].fillna("Missing name", inplace = True)
```
