# Filtrado de DataFrames con Múltiples Condiciones usando el Operador AND (`&`)

### ¿Cómo combinar múltiples criterios en Pandas?
En **Pandas**, cuando necesitamos filtrar filas que cumplan **todas** las condiciones al mismo tiempo, utilizamos el operador lógico AND a nivel de bit (`&`):

- **Uso obligatorio de paréntesis:** En Python/Pandas, los operadores de comparación (`==`, `>`, `<`, etc.) tienen menor prioridad que el operador lógico `&`. Por lo tanto, **es obligatorio encerrar cada condición individual entre paréntesis** `(condicion1) & (condicion2)`.
- **Sensibilidad a Mayúsculas/Minúsculas (*Case Sensitivity*):** Las comparaciones de cadenas de texto son exactas. Buscar `'male'` no devolverá resultados si en el DataFrame está guardado como `'Male'`.
- **Legibilidad con múltiples máscaras:** Para mantener un código limpio cuando hay muchas condiciones, se recomienda crear variables separadas (`mask1`, `mask2`, `mask3`) y luego combinarlas `df[mask1 & mask2 & mask3]`.

### Línea 12
**Código:** `mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")`

**Explicación:** Construye una máscara booleana compuesta combinando tres condiciones mediante el operador lógico AND (&). Cada condición individual debe encerrarse obligatoriamente entre paréntesis.

```python
mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")
```

### Línea 13
**Código:** `insta[mask]`

**Explicación:** Aplica la máscara booleana 'mask' para filtrar y retornar únicamente las filas del DataFrame 'insta' que cumplen las tres condiciones simultáneamente.

```python
insta[mask]
```
