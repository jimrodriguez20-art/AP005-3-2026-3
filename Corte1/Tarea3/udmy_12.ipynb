# Operaciones Aritméticas Vectoriales en Pandas

### ¿Cómo realizar cálculos matemáticos sobre columnas (Series)?
En **Pandas**, las operaciones aritméticas se ejecutan de forma **vectorial**, lo que significa que el cálculo se aplica automáticamente elemento por elemento sobre toda la columna sin necesidad de escribir bucles `for` manuales:

- **Operadores Estándar vs. Métodos de Series:** Puedes realizar operaciones usando tanto los operadores matemáticos de Python (`+`, `-`, `*`, `/`) como los métodos nativos de Pandas (`.add()`, `.sub()`, `.mul()`, `.div()`).
- **Creación de columnas calculadas:** Es posible combinar una columna existente con operaciones escalar-vectoriales (por ejemplo, convertir libras a kilos con `.mul(0.453592)`) y guardar el resultado directamente en una nueva columna.
