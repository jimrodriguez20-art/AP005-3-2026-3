# Adición, Inserción y Eliminación de Columnas en Pandas

### ¿Cómo modificar las columnas de un DataFrame?
En **Pandas**, existen distintas formas de agregar, posicionar o eliminar columnas en un **DataFrame**:

- **Adición por asignación directa (`df['Nueva_Columna'] = valor`):** Agrega una nueva columna al **final** del DataFrame con un valor constante o una secuencia de datos.
- **Inserción en una posición específica (`df.insert(loc, column, value)`):** Permite elegir la posición exacta (índice base cero) donde se ubicará la nueva columna.
- **Eliminación de columnas (`df.drop(columns, axis=1, inplace=True)`):** Remueve una o más columnas especificando `axis=1` (eje vertical/columnas). El argumento `inplace=True` modifica el DataFrame original directamente.
