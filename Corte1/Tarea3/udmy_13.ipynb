# Conteo de Frecuencias con `.value_counts()` en Pandas

### ¿Cómo contar valores únicos en una columna?
En el análisis exploratorio de datos con **Pandas**, el método `.value_counts()` sobre una **Series** (columna) permite obtener rápidamente la frecuencia con la que aparece cada valor categórico o numérico único:

- **Orden predeterminado:** Por defecto, organiza los resultados en orden descendente (de mayor a menor frecuencia).
- **Cambio de ordenación (`ascending=True`):** Permite invertir el orden del conteo para visualizar primero las categorías menos frecuentes.
- **Utilidad:** Es una de las herramientas más efectivas para analizar distribuciones de variables categóricas (como equipos, divisiones, países o géneros).
