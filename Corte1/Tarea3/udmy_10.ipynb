# Limpieza de Datos Faltantes con `.dropna()` en Pandas

### ¿Cómo eliminar filas o columnas con valores nulos (`NaN`)?
En el análisis de datos con **Pandas**, el manejo de valores faltantes es clave. El método `.dropna()` permite descartar registros incompletos de forma flexible:

- **Criterio de eliminación (`how`):** Por defecto (`how='any'`), se elimina la fila si contiene *al menos un* valor nulo. Si se usa `how='all'`, solo se elimina la fila si *todos* sus campos están vacíos.
- **Dirección de filtrado (`axis`):** Usar `axis=0` elimina **filas** (comportamiento predeterminado) y `axis=1` elimina **columnas** completas que contengan valores nulos.
- **Evaluación selectiva (`subset`):** Permite restringir la búsqueda de valores nulos a una **lista específica de columnas**, evitando eliminar filas por vacíos en columnas secundarias.
