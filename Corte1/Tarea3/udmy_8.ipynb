# Eliminación de Columnas y el Parámetro `inplace` en Pandas

### ¿Cómo eliminar columnas con el método `.drop()`?
En **Pandas**, el método `.drop()` permite remover columnas o filas de un **DataFrame** de forma sencilla:

- **Definir el eje (`axis=1`):** Para eliminar **columnas**, es obligatorio especificar `axis=1` (o `axis='columns'`), ya que por defecto `axis=0` hace referencia a filas.
- **Modificación temporal vs. permanente (`inplace`):** Por defecto, `.drop()` retorna una **copia** modificada del DataFrame sin alterar la variable original. Si se especifica `inplace=True`, los cambios se aplican de manera **permanente** directamente sobre la variable existente.
- **Eliminación múltiple:** Pasar una **lista de nombres** de columnas (ej. `['Col1', 'Col2']`) permite eliminar varias columnas en una sola operación.

### Línea 1
**Código:** `import pandas as pd`

**Explicación:** Importa la biblioteca Pandas con el alias estándar 'pd' para manejar estructuras de datos tabulares.

```python
import pandas as pd
```

### Línea 2
**Código:** `nba = pd.read_excel("Nba Teams.xlsx")`

**Explicación:** Carga el archivo Excel 'Nba Teams.xlsx' usando pd.read_excel() y asigna el DataFrame resultante a la variable 'nba'.

```python
nba = pd.read_excel("Nba Teams.xlsx")
```

### Línea 3
**Código:** `nba.head()`

**Explicación:** Visualiza en consola las primeras 5 filas del DataFrame 'nba' usando el método .head().

```python
nba.head()
```

### Línea 4
**Código:** `nba.drop("Salary", axis = 1)`

**Explicación:** Llama al método .drop() para eliminar la columna 'Salary' especificando axis=1. Devuelve una copia del DataFrame sin la columna, pero sin modificar el DataFrame 'nba' original.

```python
nba.drop("Salary", axis = 1)
```

### Línea 5
**Código:** `nba`

**Explicación:** Muestra el DataFrame 'nba' confirmando que la columna 'Salary' aún permanece en el objeto original al no haberse usado inplace=True.

```python
nba
```

### Línea 6
**Código:** `nba.drop("Salary", axis = 1, inplace = True)`

**Explicación:** Elimina la columna 'Salary' especificando axis=1 e inplace=True, lo que modifica directamente el DataFrame 'nba' en memoria sin requerir reasignación.

```python
nba.drop("Salary", axis = 1, inplace = True)
```

### Línea 7
**Código:** `nba`

**Explicación:** Visualiza el DataFrame 'nba' confirmando que la columna 'Salary' ha sido eliminada permanentemente.

```python
nba
```

### Línea 8
**Código:** `nba.drop(["Team", "Division"], axis = 1, inplace = True)`

**Explicación:** Elimina múltiples columnas al mismo tiempo ('Team' y 'Division') pasando una lista de nombres al método .drop(), con axis=1 e inplace=True para aplicar los cambios directamente.

```python
nba.drop(["Team", "Division"], axis = 1, inplace = True)
```

### Línea 9
**Código:** `nba`

**Explicación:** Muestra el estado final del DataFrame 'nba' tras la eliminación múltiple de columnas.

```python
nba
```
