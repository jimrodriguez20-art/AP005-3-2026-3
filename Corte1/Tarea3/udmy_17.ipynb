# Optimización de Datasets y Tipos de Datos en Pandas

### ¿Cómo reducir el consumo de memoria en un DataFrame?
Por defecto, al cargar archivos con **Pandas**, muchas columnas de fechas o texto repetitivo se asignan como tipo genérico `object` (cadena de texto), lo que consume una cantidad considerable de memoria RAM. Optimizar estos tipos de datos es clave para el rendimiento:

- **Conversión de Fechas (`pd.to_datetime()`):** Transforma cadenas de texto con fechas en objetos `datetime64`. Esto no solo reduce la memoria, sino que permite realizar filtrados, extracciones de año/mes y cálculos de diferencias de tiempo.
- **Conteo de Valores Únicos (`.nunique()`):** Permite verificar cuántos valores distintos existen en una columna antes de transformarla.
- **Uso de Categorías (`.astype('category')`):** Cuando una columna de texto tiene pocos valores únicos que se repiten con frecuencia (como género, país o estado), convertirla al tipo `category` reemplaza los textos redundantes por enteros internos, reduciendo el consumo de RAM hasta en un 80-90%.
