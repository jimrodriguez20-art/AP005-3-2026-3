# Ordenación de Datos con `.sort_values()` en Pandas

### ¿Cómo ordenar un DataFrame por una o más columnas?
En el análisis de datos con **Pandas**, el método `.sort_values()` permite reorganizar las filas de un **DataFrame** en función de los valores de una o varias columnas:

- **Orden Ascendente vs. Descendente (`ascending`):** Por defecto, las columnas se ordenan de forma ascendente (`ascending=True`). Para ordenar de mayor a menor o de la Z a la A, se establece `ascending=False`.
- **Ordenación por múltiples columnas:** Pasar una **lista de columnas** (ej. `['Team', 'Name']`) permite jerarquizar el ordenamiento (ordena por la primera columna y resuelve empates con la segunda).
- **Manejo de valores nulos (`na_position`):** El parámetro `na_position` permite controlar si los registros con `NaN` se ubican al final (`'last'`, por defecto) o al principio (`'first'`) del DataFrame resultantes.
