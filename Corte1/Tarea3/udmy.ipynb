# Introducción a Pandas: Carga, Inspección y Exportación de Datos

### ¿Qué es Pandas y qué es un DataFrame?
**Pandas** es una de las librerías más importantes de Python para el análisis y manipulación de datos. Su estructura fundamental es el **DataFrame**, una tabla bidimensional (organizada en filas y columnas) similar a una hoja de cálculo de Excel o a una tabla de base de datos SQL. 

En este flujo de trabajo aprenderás a importar datos desde una fuente remota (URL en formato CSV), inspeccionar sus propiedades con métodos como `.head()` e `.info()`, y filtrar/exportar los resultados a archivos locales `.csv`.

### Línea 1
**Código:** `import pandas as pd`

**Explicación:** Importa la biblioteca Pandas y le asigna el alias 'pd', el cual es el estándar para la manipulación y análisis de datos en Python.

```python
import pandas as pd
```

### Línea 2
**Código:** `url = "https://data.cityofnewyork.us/resource/rc75-m7u3.csv"`

**Explicación:** Declara la variable 'url' almacenando la dirección web del conjunto de datos CSV de casos de COVID-19 en Nueva York.

```python
url = "https://data.cityofnewyork.us/resource/rc75-m7u3.csv"
```

### Línea 3
**Código:** `#covid = pd.read_csv(url)`

**Explicación:** Comentario de código desactivado que muestra la sintaxis alternativa o previa para cargar los datos desde la URL.

### Línea 4
**Código:** `covid = pd.read_csv(url)`

**Explicación:** Llama a la función pd.read_csv() pasando la URL como argumento para descargar y convertir los datos remotos en un DataFrame asignado a 'covid'.

```python
covid = pd.read_csv(url)
```

### Línea 5
**Código:** `covid.head()`

**Explicación:** Invoca el método .head() sobre el DataFrame 'covid' para visualizar en consola las primeras 5 filas del conjunto de datos.

```python
covid.head()
```

### Línea 6
**Código:** `covid.info()`

**Explicación:** Llama al método .info() para mostrar un resumen estructural del DataFrame (cantidad de filas, columnas, tipos de datos y memoria utilizada).

```python
covid.info()
```

### Línea 7
**Código:** `covid.to_csv('NYC_covid_cases.csv')`

**Explicación:** Exporta el DataFrame completo a un archivo local de texto plano con formato CSV llamado 'NYC_covid_cases.csv'.

```python
covid.to_csv('NYC_covid_cases.csv')
```

### Línea 8
**Código:** `covid.to_csv('NYC1_covid_cases.csv', columns = ['case_count', 'hospitalized_count', 'death_count'])`

**Explicación:** Exporta únicamente un subconjunto específico de columnas ('case_count', 'hospitalized_count' y 'death_count') a un nuevo archivo CSV llamado 'NYC1_covid_cases.csv'.

```python
covid.to_csv('NYC1_covid_cases.csv', columns = ['case_count', 'hospitalized_count', 'death_count'])
```
