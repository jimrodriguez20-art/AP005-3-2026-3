# Creación de DataFrames a partir de Diccionarios en Pandas

### ¿Cómo se crea un DataFrame desde un diccionario?
En **Pandas**, una de las formas más intuitivas y comunes de construir un **DataFrame** es a partir de un **diccionario de listas**.

- **Claves del diccionario:** Se convierten automáticamente en las **columnas** de la tabla.
- **Listas de valores:** Cada elemento de la lista pasa a ser una **fila** dentro de su respectiva columna (todas las listas deben tener la misma longitud).
- **Índices de fila:** Por defecto se asignan valores numéricos comenzando en `0`, pero pueden personalizarse utilizando el parámetro `index`.

### Línea 1
**Código:** `import pandas as pd`

**Explicación:** Importa la biblioteca Pandas bajo el alias estándar 'pd' para manejar estructuras de datos tabulares.

```python
import pandas as pd
```

### Línea 2
**Código:** `data = {"name": ["Sally", "Jhon", "Kim"], "age": [4, 6 ,7], "gender": ["female", "male", "male"]}`

**Explicación:** Declara la variable 'data' asignando un diccionario donde las claves representan los nombres de las columnas y las listas de valores representan los datos de cada columna.

```python
data = {"name": ["Sally", "Jhon", "Kim"], "age": [4, 6 ,7], "gender": ["female", "male", "male"]}
```

### Línea 3
**Código:** `pd.DataFrame(data)`

**Explicación:** Construye un DataFrame a partir del diccionario 'data'. Asigna automáticamente las claves del diccionario como encabezados de columna e índices numéricos secuenciales (0, 1, 2) a las filas.

```python
pd.DataFrame(data)
```

### Línea 4
**Código:** `pd.DataFrame(data, index = ["Kid1", "Kid2", "Kid3"])`

**Explicación:** Crea el DataFrame especificando etiquetas personalizadas para los índices de las filas mediante el parámetro 'index' ('Kid1', 'Kid2', 'Kid3').

```python
pd.DataFrame(data, index = ["Kid1", "Kid2", "Kid3"])
```
