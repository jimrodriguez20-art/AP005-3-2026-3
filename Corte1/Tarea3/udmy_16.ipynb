# Encadenamiento de Métodos en Carga e Imputación de Columnas en Pandas

### ¿Cómo optimizar la carga de datos y el manejo de nulos?
En **Pandas**, es posible combinar múltiples operaciones en una sola instrucción mediante la técnica de **method chaining** (encadenamiento de métodos):

- **Limpieza al cargar (`pd.read_excel().dropna()`):** Permite filtrar registros indeseados (como filas totalmente vacías con `how='all'`) en el mismo instante en que se lee el archivo, evitando almacenar datos innecesarios en memoria.
- **Diagnóstico directo (`.info()`):** Permite verificar la eficacia de la limpieza inicial revisando la cantidad de registros no nulos restantes por columna.
- **Imputación de cadenas (`.fillna()`):** Reemplaza valores ausentes en columnas de texto o categóricas por identificadores claros como `'Missing'` para mantener la consistencia en el análisis.
