# Creación de DataFrames a partir de Listas en Pandas

### ¿Cómo se crea un DataFrame desde una lista?
En **Pandas**, el constructor `pd.DataFrame()` permite transformar estructuras de datos nativas de Python (como listas simples o anidadas) en tablas de datos organizadas.

- **Listas unidimensionales:** Se convierten en DataFrames de una sola columna.
- **Listas bidimensionales (listas de listas):** Cada sublista se interpreta como una fila de la tabla.
- **Personalización:** Es posible definir los encabezados de las columnas mediante el parámetro `columns` y las etiquetas de las filas utilizando el parámetro `index`.

**Explicación:** Importa la biblioteca Pandas bajo el alias estándar 'pd' para habilitar sus estructuras de datos y funciones.

```python
import pandas as pd
```

### Línea 3
**Código:** `numbers = [1, 2, 3, 4, 5, 6]`

**Explicación:** Declara la variable 'numbers' e inicializa una lista unidimensional de números enteros.

```python
numbers = [1, 2, 3, 4, 5, 6]
```

### Línea 4
**Código:** `pd.DataFrame(numbers)`

**Explicación:** Llama al constructor pd.DataFrame() pasando la lista simple 'numbers'. Crea una estructura tabular de una sola columna con índices numéricos por defecto (0 a 5).

```python
pd.DataFrame(numbers)
```

### Línea 5
**Código:** `data = [["Jhon", 23], ["Mary", 18], ["Jhon", 31]]`

**Explicación:** Declara la variable 'data' como una lista bidimensional (lista de listas), donde cada sublista representa una fila con valores de nombre y edad.

```python
data = [["Jhon", 23], ["Mary", 18], ["Jhon", 31]]
```

### Línea 6
**Código:** `pd.DataFrame(data)`

**Explicación:** Convierte la lista anidada 'data' en un DataFrame. Asigna nombres automáticos a las columnas (0 y 1) e índices numéricos a las filas (0, 1, 2).

```python
pd.DataFrame(data)
```

### Línea 7
**Código:** `pd.DataFrame(data, columns = ["name", "age"])`

**Explicación:** Construye el DataFrame asignando nombres explícitos a las columnas mediante el argumento nombrado 'columns' (columna 'name' y columna 'age').

```python
pd.DataFrame(data, columns = ["name", "age"])
```

### Línea 8
**Código:** `pd.DataFrame(data, columns = ["name", "age"], index = ["id1", "id2", "id3"])`

**Explicación:** Crea el DataFrame especificando tanto los nombres de las columnas ('columns') como identificadores personalizados para las filas mediante el argumento 'index'.

```python
pd.DataFrame(data, columns = ["name", "age"], index = ["id1", "id2", "id3"])
```
