## 1. Creación básica de un diccionario

Un **diccionario** en Python es una estructura de datos que almacena pares **llave-valor** (*key-value pairs*). A diferencia de una lista, en la que se accede a los elementos por su posición (índice), en un diccionario se accede a los valores usando su llave.

En este primer bloque se crean dos diccionarios simples: uno que asocia habitaciones de una casa con lecturas de un sensor (por ejemplo, temperatura), y otro que asocia zonas de una propiedad con el número de cámaras instaladas.

````python
# Diccionario que asocia cada habitación con una lectura numérica (por ejemplo, temperatura)
sensors = {"living room": 21, "kitchen": 23, "bedroom": 20, "pantry": 22}

# Diccionario que asocia cada zona con la cantidad de cámaras instaladas
num_cameras = {"backyard": 6, "garage": 2, "driveway": 1}

print(sensors)      # Imprime el diccionario completo de sensores
print(num_cameras)  # Imprime el diccionario completo de cámaras
````

## 2. Diccionarios con valores de tipo `string`

Los valores de un diccionario no tienen que ser números: pueden ser cadenas de texto, listas, otros diccionarios, etc. Aquí se crea un diccionario de "traducciones" donde tanto la llave como el valor son `strings`.

````python
# Diccionario de traducciones: palabra en inglés -> palabra inventada/ficticia
translations = {"mountain": "orod", "bread": "bass", "friend": "mellon", "horse": "roch"}

print(translations)  # Imprime el diccionario completo de traducciones
````

## 3. Punto de atención: las llaves deben ser *hashables* (error real)

Las llaves de un diccionario **deben ser de un tipo inmutable** (técnicamente, *hashable*): `str`, `int`, `float`, `tuple`, entre otros. Una **lista es mutable**, por lo tanto **no puede usarse como llave** de un diccionario.

La siguiente línea, si se ejecuta, produce:

```
TypeError: unhashable type: 'list'
```

Se deja comentada intencionalmente para que el notebook completo pueda ejecutarse sin interrupciones, pero se incluye como ejemplo de un error común.

````python
# ADVERTENCIA: la siguiente línea produce un error si se descomenta y ejecuta.
# Las listas son mutables, así que no son "hashables" y no pueden usarse como llaves de un diccionario.
# powers = {[1, 2, 4, 8, 16]: 2, [1, 3, 9, 27, 81]: 3}

# Si se descomenta la línea anterior, Python lanzará:
# TypeError: unhashable type: 'list'
````

## 4. Listas como *valores* del diccionario (esto sí es válido)

A diferencia de las llaves, los **valores** de un diccionario sí pueden ser listas (o cualquier otro tipo, incluso mutable). Aquí cada apellido de familia se asocia con una lista de nombres.

````python
# Cada llave (apellido de familia) se asocia con una lista de nombres (valor)
children = {"von Trapp": ["Johannes", "Rosmarie", "Eleonore"], "Corleone": ["Sonny", "Fredo", "Michael"]}

print(children)  # Imprime el diccionario con listas como valores
````

## 5. Diccionario vacío

Se puede crear un diccionario vacío con `{}`. Esto es útil cuando se planea llenarlo dinámicamente más adelante, por ejemplo dentro de un bucle.

````python
my_empty_dictionary = {}  # Crea un diccionario vacío
print(my_empty_dictionary)  # Imprime: {}
````

## 6. Agregar una nueva llave a un diccionario existente

Para agregar una nueva llave (con su valor) a un diccionario ya existente, se usa la sintaxis `diccionario["nueva_llave"] = valor`.

````python
menu = {"oatmeal": 3, "avocado toast": 6, "carrot juice": 5, "blueberry muffin": 2}
print("Before: ", menu)  # Estado del menú antes de agregar el nuevo ítem

menu["cheesecake"] = 8  # Agrega la llave "cheesecake" con valor 8 al diccionario
print("After", menu)    # Estado del menú después de agregar el nuevo ítem
````

## 7. Punto de atención: reasignar una variable no es lo mismo que actualizar un diccionario

En el siguiente bloque, la variable `animals_in_zoo` se **reasigna por completo** tres veces seguidas, en vez de **actualizarse**. Esto significa que cada nueva asignación **reemplaza** todo el diccionario anterior, en lugar de agregarle una llave nueva.

Como resultado, la información de `"dinosaurs"` se pierde y solo queda `{"horses": 2}`. Este comportamiento es correcto desde el punto de vista de Python (no es un error de sintaxis), pero se señala aquí porque suele ser una fuente común de confusión: si la intención hubiera sido acumular animales, habría que usar `.update()` o `animals_in_zoo["horses"] = 2` en vez de una reasignación completa.

````python
animals_in_zoo = {"dinosaurs": 0}
animals_in_zoo = {"dinosaurs": 0}  # Reasignación redundante: vuelve a crear el mismo diccionario
animals_in_zoo = {"horses": 2}     # OJO: esto NO actualiza el diccionario, lo REEMPLAZA por completo

print(animals_in_zoo)  # Solo imprime {"horses": 2}; la llave "dinosaurs" ya no existe
````

## 8. Agregar múltiples llaves con `.update()`

El método `.update()` permite agregar (o modificar) varias llaves de un diccionario a la vez, pasándole otro diccionario como argumento.

````python
sensors = {"living room": 21, "kitchen": 23, "bedroom": 20}
print("Before", sensors)  # Estado antes de agregar nuevas habitaciones

# Agrega 3 llaves nuevas (pantry, guest room, patio) en una sola operación
sensors.update({"pantry": 22, "guest room": 25, "patio": 34})
print("After", sensors)  # Estado después de agregar las nuevas habitaciones
````

## 9. Otro ejemplo de `.update()`

Un segundo ejemplo del mismo patrón, esta vez con identificadores de usuario.

````python
user_ids = {"teraCoder": 9018293, "proProgrammer": 119238}
print(user_ids)  # Diccionario original de IDs de usuario

# Agrega dos nuevos usuarios al diccionario
user_ids.update({"theLooper": 138475, "stringQueen": 85739})
print(user_ids)  # Diccionario con los nuevos usuarios agregados
````

## 10. Sobrescribir el valor de una llave existente

Si se usa `diccionario["llave"] = nuevo_valor` sobre una llave que **ya existe**, el valor anterior se sobrescribe (no se crea una llave nueva ni se duplica).

````python
menu = {"oatmeal": 3, "avocado toast": 6, "carrot juice": 5, "blueberry muffin": 2}
print("Before: ", menu)  # Precio original de "oatmeal": 3

menu["oatmeal"] = 5  # Sobrescribe el valor de "oatmeal" (de 3 pasa a 5)
print("After", menu)  # El precio de "oatmeal" ahora es 5
````

## 11. Comparando `.update()` con la asignación directa

Este ejemplo combina los dos mecanismos vistos: `.update()` para **agregar** una llave nueva, y la asignación directa (`diccionario["llave"] = valor`) para **sobrescribir** una llave existente.

````python
oscar_winners = {"Best Picture": "La La Land", "Best Actor": "Casey Affleck", "Best Actress": "Emma Stone", "Animated Feature": "Zootopia"}
print("Before", oscar_winners)
print()

oscar_winners.update({"Supporting Actress": "Viola Davis"})  # Agrega una llave nueva
print("After1", oscar_winners)
print()

oscar_winners["Best Picture"] = "Moonlight"  # Sobrescribe el valor existente de "Best Picture"
print("After2", oscar_winners)
````

## 12. `zip()` y *dict comprehensions*

La función `zip()` combina dos (o más) listas en un **iterador de tuplas**, emparejando los elementos por posición. Es importante notar que al imprimir directamente el resultado de `zip()` **no se ven las tuplas**, sino la representación del objeto iterador (algo como `<zip object at 0x...>`). Para ver el contenido hay que recorrerlo (por ejemplo, convirtiéndolo en `list()` o usándolo en un bucle/comprehension).

Una **dict comprehension** permite construir un diccionario a partir de ese iterador de pares, con la sintaxis `{clave: valor for clave, valor in iterable}`.

````python
names = ['Jenny', 'Alexus', 'Sam', 'Grace']
heights = [61, 70, 67, 64]

zipStudents = zip(names, heights)  # Combina ambas listas en un iterador de tuplas (nombre, altura)
print("zipStudents: ", zipStudents)  # OJO: esto imprime el objeto iterador, no su contenido (ej. <zip object at 0x...>)
````

````python
# Dict comprehension: recorre los pares (nombre, altura) y construye el diccionario
students = {key: value for key, value in zip(names, heights)}
# students queda como: {'Jenny': 61, 'Alexus': 70, 'Sam': 67, 'Grace': 64}

print(students)  # Imprime el diccionario ya construido
````

## 13. Otro ejemplo de `zip()` + dict comprehension

Se repite el mismo patrón con una lista de bebidas y su contenido de cafeína.

````python
drinks = ["espresso", "chai", "decaf", "drip"]
caffeine = [64, 40, 0, 120]

zipped_drinks = zip(drinks, caffeine)  # Iterador de tuplas (bebida, cantidad de cafeína)
print(zipped_drinks)  # Nuevamente se imprime el objeto iterador, no las tuplas

drinks_to_caffeine = {key: value for key, value in zipped_drinks}  # Construye el diccionario final
print(drinks_to_caffeine)  # Imprime el diccionario bebida -> cafeína
````

## 14. Ejemplo final: diccionarios anidados

Este último bloque combina varios conceptos ya vistos: creación de un diccionario con dict comprehension, actualización con `.update()` (agregando una llave nueva y sobrescribiendo una existente), y finalmente la creación de un **diccionario anidado**, donde uno de los valores es a su vez otro diccionario.

````python
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]
playcounts = [78, 29, 44, 21, 89, 5]

# Crea el diccionario canción -> número de reproducciones usando zip() + dict comprehension
plays = {key: value for key, value in zip(songs, playcounts)}
print(plays)

plays.update({"Purple Haze": 1})  # Agrega una canción nueva que no estaba en la lista original
plays.update({"Respect": 94})     # Sobrescribe el número de reproducciones de "Respect" (de 89 pasa a 94)
print("After: ", plays)

# Diccionario anidado: sus valores son a su vez otros diccionarios
library = {"The Best Songs": plays, "Sunday Feelings": {}}
print(library)
````

## Conclusión

En este notebook se repasaron los conceptos fundamentales sobre **diccionarios** en Python:

- Un diccionario almacena pares **llave-valor**, y se accede a los valores mediante su llave (no por posición, como en una lista).
- Las **llaves** deben ser de un tipo inmutable (*hashable*): usar una lista como llave produce un `TypeError`. Los **valores**, en cambio, pueden ser de cualquier tipo, incluyendo listas u otros diccionarios (diccionarios anidados).
- Se puede **agregar o sobrescribir** una sola llave con `diccionario["llave"] = valor`, o **agregar/actualizar varias llaves a la vez** con `.update()`.
- **Reasignar** una variable (`diccionario = {...}`) no es lo mismo que **actualizarla**: la reasignación reemplaza todo el contenido anterior, mientras que `.update()` o la asignación por llave conservan el resto de los datos.
- La función `zip()` combina listas en un iterador de tuplas, que puede transformarse en un diccionario mediante una **dict comprehension** (`{clave: valor for clave, valor in iterable}`). Al imprimir un objeto `zip` directamente se ve el iterador, no su contenido.

Estos conceptos son la base para trabajar con estructuras de datos más complejas en Python, como diccionarios anidados o combinaciones de diccionarios y listas.
